```python
!pip install -q networkx

import networkx as nx
import blackhole_diffusion as bd

G = nx.erdos_renyi_graph(10, 0.4, seed=42)
edges = set(G.edges())

cycles = bd.chordless_cycle_basis(edges)

expected = len(edges) - len(G.nodes) + 1
print(f"V={len(G.nodes)}, E={len(edges)}")
print(f"Expected: {expected}, Got: {len(cycles)}")
print("OK" if len(cycles) == expected else "MISMATCH")
print(f"All cycles >= 3: {all(len(c) >= 3 for c in cycles)}")
```
