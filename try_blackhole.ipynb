```python
!pip install -q networkx   # networkx 仅用于生成测试图，你的算法不依赖它

import networkx as nx
import blackhole_diffusion as bd

# 生成一个小型连通随机图（10 节点，约 40% 边概率）
G = nx.erdos_renyi_graph(10, 0.4, seed=42)
edges = set(G.edges())          # 边集：Set[Tuple[int, int]]

# 调用你的公开 API —— 顶点集合版
cycles = bd.chordless_cycle_basis(edges)

# 环基的秩应该等于 E - V + 1（对于连通图）
expected = len(edges) - len(G.nodes) + 1
print(f"Graph: V={len(G.nodes)}, E={len(edges)}")
print(f"Expected cycle basis rank: {expected}")
print(f"Returned number of cycles: {len(cycles)}")

if len(cycles) == expected:
    print("✓ Rank matches the cyclomatic number.")
else:
    print(f"✗ Rank mismatch (got {len(cycles)}, expected {expected})")

# 再检查一下每个环是否至少包含 3 个顶点
all_ok = all(len(c) >= 3 for c in cycles)
print(f"All cycles have length ≥ 3: {all_ok}")
```
